```python
# Correlation Analysis
df.corr(numeric_only=True)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Correlation Heatmap
plt.figure(figsize=(8,5))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

#### Observation:

Red / warm colors → positive correlation

Blue / cool colors → negative correlation

White/light → weak or no correlation

```python
#Readmission countplot
sns.countplot(x="Readmission", data=df)
plt.show()
```

#### Observation:
- Less people were readmitted
- about +2000 people were not readmitted

```python
df["Age"].hist()

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.show()
```

```python
df["Gender"].value_counts().plot(kind="bar")

plt.title("Diagnoses Count")
plt.xlabel("Gender")
plt.ylabel("Number of Diagnoses")

plt.show()
```

#### Observation:

- Males had slightly more diagnoses than females

```python
import matplotlib.pyplot as plt

df["Age"].hist(bins=10)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```
